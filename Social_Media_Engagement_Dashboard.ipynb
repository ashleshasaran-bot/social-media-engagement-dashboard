# Social Media Engagement Dashboard

**Project Type:** Data Analytics  
**Goal:** Analyse social media post performance using likes, comments, shares and views.


## 1. Business Problem

Analyse which social media platforms and post types generate the strongest engagement.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Create Sample Social Media Data


```python
data = {
'Platform':['Instagram','LinkedIn','Facebook','X','Instagram','LinkedIn','Facebook','X','Instagram','LinkedIn','Facebook','X'],
'Post_Type':['Reel','Article','Image','Text','Image','Video','Video','Image','Reel','Article','Image','Text'],
'Likes':[520,340,280,190,610,430,350,210,700,390,320,230],
'Comments':[48,62,35,28,55,70,42,30,68,58,40,35],
'Shares':[80,95,55,40,110,120,65,45,130,105,60,50],
'Views':[5200,3100,4200,2900,6400,3900,5100,3200,7200,3500,4700,3000]
}
df = pd.DataFrame(data)
display(df)
```

## 3. Calculate Engagement Metrics

**Total Engagement = Likes + Comments + Shares**

**Engagement Rate = Total Engagement ÷ Views × 100**


```python
df['Total_Engagement'] = df['Likes'] + df['Comments'] + df['Shares']
df['Engagement_Rate'] = (df['Total_Engagement'] / df['Views'] * 100).round(2)
display(df)
```

## 4. Dashboard Summary


```python
summary = pd.DataFrame({
'Metric':['Total Posts','Total Likes','Total Comments','Total Shares','Total Views','Average Engagement Rate'],
'Value':[len(df),df['Likes'].sum(),df['Comments'].sum(),df['Shares'].sum(),df['Views'].sum(),round(df['Engagement_Rate'].mean(),2)]
})
display(summary)
```

## 5. Platform Performance


```python
platform_performance = df.groupby('Platform').agg(
Average_Likes=('Likes','mean'),
Average_Comments=('Comments','mean'),
Average_Shares=('Shares','mean'),
Average_Engagement_Rate=('Engagement_Rate','mean')
).round(2)
display(platform_performance)
```

```python
plt.figure(figsize=(8,4))
df.groupby('Platform')['Total_Engagement'].mean().sort_values(ascending=False).plot(kind='bar')
plt.title('Average Total Engagement by Platform')
plt.xlabel('Platform')
plt.ylabel('Average Engagement')
plt.xticks(rotation=0)
plt.show()
```

## 6. Engagement by Post Type


```python
post_type = df.groupby('Post_Type')['Total_Engagement'].mean().sort_values(ascending=False)
display(post_type.to_frame(name='Average Engagement'))

plt.figure(figsize=(8,4))
post_type.plot(kind='bar')
plt.title('Average Engagement by Post Type')
plt.xlabel('Post Type')
plt.ylabel('Average Engagement')
plt.xticks(rotation=0)
plt.show()
```

## 7. Top Performing Posts


```python
top_posts = df.sort_values('Engagement_Rate', ascending=False).head(5)
display(top_posts)
```
